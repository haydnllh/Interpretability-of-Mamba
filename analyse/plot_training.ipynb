```python
from tensorboard.backend.event_processing import event_accumulator
import matplotlib.pyplot as plt

# Load the TensorBoard event file
ea = event_accumulator.EventAccumulator('/scratch/lhl1g23/mamba/lightning_logs/wikitext-103/gpt2/2nd/version_0')
ea.Reload()

# Extract a scalar (e.g. training loss)
loss = ea.Scalars('train_loss_epoch')

steps = [s.step for s in loss]
values = [s.value for s in loss]
print(values[-1])

plt.plot(steps, values)
plt.xlabel('Step')
plt.ylabel('Loss')
plt.title('Validation Loss')
plt.grid(alpha=0.5)
plt.show()
```
